# Static Link Prediction benchmark results

Loads `data/static_link_prediction.csv` (produced by
`scripts/harness/benchmark_tasks/parse_benchmark_eval_logs.py`) and summarises
**ROC-AUC** (held-out edges vs. 2-hop hard negatives) per (model × dataset).
Accuracy/F1 are also in the CSV.

Regenerate the CSV from Tucker logs, e.g.:
```
python scripts/harness/benchmark_tasks/parse_benchmark_eval_logs.py \
    --log-root /dataMeR2/phil/gfm/prodigy/log --out-dir scripts/experiments/analysis
```

````python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

csv = Path('data/static_link_prediction.csv')
df = pd.read_csv(csv)
df = df[df['split'] == 'test'].copy()
print(df.shape)
df.head()
````

````python
# ROC-AUC by model x dataset (best checkpoint per cell)
pivot = df.pivot_table(index='model', columns='dataset', values='roc_auc', aggfunc='max')
pivot
````

````python
ax = pivot.plot(kind='bar', figsize=(10, 5))
ax.set_ylabel('ROC-AUC (test, hard negatives)')
ax.set_title('Static link prediction by model and dataset')
ax.axhline(0.5, color='k', lw=0.6, ls='--', label='chance')
ax.set_ylim(0.4, 1.0)
ax.legend(title='dataset', bbox_to_anchor=(1.01, 1), loc='upper left')
plt.tight_layout()
plt.show()
````
