# Node Regression benchmark results

Loads `data/node_regression.csv` (produced by
`scripts/harness/benchmark_tasks/parse_benchmark_eval_logs.py`) and summarises
**Spearman ρ** per (model × dataset × profile target). Spearman is the headline
metric; RMSE/MAE/R² are also in the CSV.

Regenerate the CSV from Tucker logs, e.g.:
```
python scripts/harness/benchmark_tasks/parse_benchmark_eval_logs.py \
    --log-root /dataMeR2/phil/gfm/prodigy/log --out-dir scripts/experiments/analysis
```

````python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

csv = Path('data/node_regression.csv')
df = pd.read_csv(csv)
df = df[df['split'] == 'test'].copy()
print(df.shape)
df.head()
````

````python
# Spearman by dataset x target (best checkpoint per cell)
pivot = df.pivot_table(index='dataset', columns='target', values='spearman', aggfunc='max')
pivot
````

````python
ax = pivot.plot(kind='bar', figsize=(10, 5))
ax.set_ylabel('Spearman ρ (test)')
ax.set_title('Node regression: rank correlation by dataset and target')
ax.axhline(0, color='k', lw=0.6)
ax.legend(title='target', bbox_to_anchor=(1.01, 1), loc='upper left')
plt.tight_layout()
plt.show()
````
